```python
import math
import matplotlib.pyplot as plt
from scipy import stats
time=[15, 29,8,42,35,21,18,42,26]
mean=sum(time)/len(time)
print(mean)
median=len(time)/2
print(median)
Max_count=0
mode=None
for num in time:
   if time.count(num)>Max_count:
     Max_count=time.count(num)
     mode=num
print(mode)
squared_differences = [(x - mean) ** 2 for x in time]
variance = sum(squared_differences) / (len(time) -1)
print(variance)
standard_deviation= math.sqrt(variance)
print(standard_deviation)
range_val = max(time) - min(time)
print(range_val)
```

```python
def factorial_iterative(n):
    if n < 0:
        return "Factorial does not exist for negative numbers"
    
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
num = int(input("Enter a number: "))
print(f"The factorial of {num} is {factorial_iterative(num)}")
```

```python
def sum_of_digits(n):
    n = abs(n)
    total_sum = 0
    
    while n > 0:
        digit = n % 10
        total_sum += digit
        n = n // 10
        
    return total_sum
num = int(input("Enter a number: "))
result = sum_of_digits(num)
print(f"The sum of the digits is: {result}")
```

```python
import math

def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    
    for i in range(3, int(math.sqrt(n)) + 1, 2):
        if n % i == 0:
            return False
            
    return True

num = int(input("Enter a number: "))
if is_prime(num):
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```

```python
def is_perfect_number(n):
    if n <= 0:
        return False
    
    divisor_sum = 0
    for i in range(1, n):
        if n % i == 0:
            divisor_sum += i
            
    return divisor_sum == n

num = int(input("Enter an integer: "))

if is_perfect_number(num):
    print(f"{num} is a perfect number.")
else:
    \
    print(f"{num} is not a perfect number.")
    
```

```python
def generate_fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    fib_sequence = [0, 1]
    while len(fib_sequence) < n:
        next_value = fib_sequence[-1] + fib_sequence[-2]
        fib_sequence.append(next_value)
        
    return fib_sequence
count = int(input("How many Fibonacci numbers would you like? "))
result = generate_fibonacci(count)

print(f"The first {count} numbers are: {result}")
```

```python
def is_armstrong(n):
    num_str = str(n)
    num_digits = len(num_str)
    total_sum = sum(int(digit) ** num_digits for digit in num_str)
    
    return total_sum == n
num = int(input("Enter an integer: "))

if is_armstrong(num):
    print(f"{num} is an Armstrong number.")
else:
    print(f"{num} is not an Armstrong number.")
```
