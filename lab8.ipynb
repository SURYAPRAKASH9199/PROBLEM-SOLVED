```python
#1. Compute mean, median, mode, variance and standard deviation of a data column. use cleaned_data.csv with coloumn names Car_Name,Year,Selling_Price,Present_Price,Kms_Driven,Fuel_Type,Seller_Type,Transmission,Owner
import numpy as np
import pandas as pd
from scipy import stats
# Load the dataset
data = pd.read_csv(r"C:\Users\KIIT0001\Downloads\car data.csv")
# Select the numeric column for analysis
numeric_column = data['Selling_Price']  # Replace with your desired column name
# Calculate mean, median, mode, variance, and standard deviation
mean = np.mean(numeric_column)
median = np.median(numeric_column)
mode = stats.mode(numeric_column).mode
variance = np.var(numeric_column, ddof=1)  # Sample variance
std_dev = np.std(numeric_column, ddof=1)  # Sample standard deviation
# Print the results
print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Mode: {mode}")
print(f"Variance: {variance}")
print(f"Standard Deviation: {std_dev}")


#2. Standardize a numeric column (z-score normalization).
# Select the numeric column for standardization
numeric_column = data['Selling_Price']  # Replace with your desired column name
# Calculate mean and standard deviation
mean = np.mean(numeric_column)
std_dev = np.std(numeric_column, ddof=1)  # Sample standard deviation
# Perform z-score normalization
z_score_normalized = (numeric_column - mean) / std_dev
# Print the standardized values
print(z_score_normalized)

#3. Apply min-max normalization on a dataset column.
# Select the numeric column for min-max normalization
numeric_column = data['Selling_Price']  # Replace with your desired column name
# Calculate min and max values
min_val = numeric_column.min()
max_val = numeric_column.max()
# Perform min-max normalization
min_max_normalized = (numeric_column - min_val) / (max_val - min_val)
# Print the normalized values
print(min_max_normalized)

#4. Create a random sample of size 100 from a normally distributed population.
# Set the parameters for the normal distribution
mean = 0  # Mean of the distribution
std_dev = 1  # Standard deviation of the distribution
# Generate a random sample of size 100
random_sample = np.random.normal(mean, std_dev, 100)
# Print the random sample
print(random_sample)

#5. Write a program to calculate conditional probability P(A|B) from given frequencies.
# Define the frequencies
frequency_A_and_B = 30  # Frequency of both A and B occurring
frequency_B = 50  # Frequency of B occurring
# Calculate conditional probability P(A|B)
if frequency_B > 0:
    conditional_probability = frequency_A_and_B / frequency_B
    print(f"P(A|B) = {conditional_probability}")
else:
    print("Frequency of B cannot be zero.")

#6. Compute correlation coefficient between two numeric attributes.
# Select two numeric columns for correlation analysis
numeric_column_1 = data['Selling_Price']  # Replace with your desired column name
numeric_column_2 = data['Present_Price']  # Replace with your desired column name
# Calculate the correlation coefficient
correlation_coefficient = np.corrcoef(numeric_column_1, numeric_column_2)[0, 1]
# Print the correlation coefficient
print(f"Correlation Coefficient: {correlation_coefficient}")

#7. Perform train-test split manually without using sklearn.
# Define the dataset and the split ratio
dataset = data  # Replace with your dataset
split_ratio = 0.8  # 80% for training, 20% for testing
# Shuffle the dataset
shuffled_dataset = dataset.sample(frac=1, random_state=42).reset_index(drop=True)
# Calculate the split index
split_index = int(len(shuffled_dataset) * split_ratio)
# Split the dataset into training and testing sets
train_set = shuffled_dataset[:split_index]
test_set = shuffled_dataset[split_index:]
# Print the sizes of the training and testing sets
print(f"Training Set Size: {len(train_set)}")
print(f"Testing Set Size: {len(test_set)}")

#8. Convert a skewed column using log-transformation and analyze improvement
# Select the skewed column for log-transformation
skewed_column = data['Selling_Price']  # Replace with your desired column name
# Apply log-transformation
log_transformed = np.log(skewed_column + 1)  # Adding 1 to avoid log(0)
# Analyze improvement by comparing skewness before and after transformation
skewness_before = skewed_column.skew()
skewness_after = log_transformed.skew()
print(f"Skewness before log-transformation: {skewness_before}")
print(f"Skewness after log-transformation: {skewness_after}")
```
