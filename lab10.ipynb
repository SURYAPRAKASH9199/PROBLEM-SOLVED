```python
X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]

n = len(X)

# Calculate means
mean_x = sum(X) / n
mean_y = sum(Y) / n

# Calculate slope (B1)
numerator = 0
denominator = 0

for i in range(n):
    numerator += (X[i] - mean_x) * (Y[i] - mean_y)
    denominator += (X[i] - mean_x) ** 2

B1 = numerator / denominator

# Calculate intercept (B0)
B0 = mean_y - B1 * mean_x

# Print regression equation
print("Regression Equation:")
print(f"y = {B0:.2f} + {B1:.2f}x")

# Prediction for a given input
x_input = float(input("Enter value of x for prediction: "))
y_pred = B0 + B1 * x_input

print(f"Predicted value of y: {y_pred:.2f}")
```

```python
import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]

n = len(X)

B0 = 0
B1 = 0

alpha = 0.01
iterations = 100

cost_list = []

for it in range(iterations):
    sum_error = 0
    dB0 = 0
    dB1 = 0

    for i in range(n):
        y_pred = B0 + B1 * X[i]
        error = y_pred - Y[i]
        sum_error += error ** 2
        dB0 += error
        dB1 += error * X[i]

    cost = sum_error / (2 * n)
    cost_list.append(cost)

    B0 = B0 - alpha * (dB0 / n)
    B1 = B1 - alpha * (dB1 / n)

print(f"y = {B0:.2f} + {B1:.2f}x")

plt.plot(range(iterations), cost_list)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Cost vs Iterations")
plt.show()
```

```python
import numpy as np

X = np.array([[1200, 3],
              [1500, 4],
              [1800, 5],
              [2000, 5]])

Y = np.array([300000, 360000, 400000, 450000])

ones = np.ones((X.shape[0], 1))
X = np.hstack((ones, X))

beta = np.linalg.inv(X.T @ X) @ X.T @ Y

test = np.array([1, 1600, 4])
prediction = test @ beta

print("Coefficients:", beta)
print("Prediction:", prediction)
```
