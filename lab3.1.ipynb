```python
numbers = [23, 1, 45, 98, 12, 5, 67]

if not numbers:
    print("The list is empty.")
else:
    max_val = numbers[0]
    min_val = numbers[0]

    for num in numbers:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num

    print(f"List: {numbers}")
    print(f"Maximum: {max_val}")
    print(f"Minimum: {min_val}")
```
