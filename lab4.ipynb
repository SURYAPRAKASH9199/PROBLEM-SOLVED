```python
def find_extremes(numbers):
    if not numbers:
        return None, None
    current_max = numbers[0]
    current_min = numbers[0]
    for num in numbers[1:]:
        if num > current_max:
            current_max = num
        if num < current_min:
            current_min = num
    return current_max, current_min
my_list = [23, 5, 89, 2, 45, 100, -3]
max_val, min_val = find_extremes(my_list)

print(f"The maximum value is: {max_val}")
print(f"The minimum value is: {min_val}")
```

```python
def count_even_odd(numbers):
    even_count = 0
    odd_count = 0    
    for num in numbers:
        if num % 2 == 0:
            even_count += 1
        else:
            odd_count += 1            
    return even_count, odd_count
my_list = [10, 21, 4, 45, 66, 93, 11]
evens, odds = count_even_odd(my_list)
print("Even numbers:", evens)
print("Odd numbers:", odds)
```

```python
def remove_duplicates(input_list):
    unique_list = []
    
    for item in input_list:
        if item not in unique_list:
            unique_list.append(item)            
    return unique_list
my_list = [1, 2, 2, 3, 4, 4, 4, 5, 1]
result = remove_duplicates(my_list)
print("Original list:", my_list)
print("List after removing duplicates:", result)
```

```python
def find_second_largest(numbers):
    if len(numbers) < 2:
        return None
    first = second = float('-inf')

    for num in numbers:
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
    if second == float('-inf'):
        return None
    
    return second
my_list = [10, 20, 4, 45, 99, 99, 34]
result = find_second_largest(my_list)
print("The second largest element is:", result)
```

```python
def bubble_sort(arr):
    n = len(arr)
    
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                
    return arr

my_list = [64, 34, 25, 12, 22, 11, 90]
sorted_list = bubble_sort(my_list)

print("Sorted list:", sorted_list)
```

```python
def merge_lists(list1, list2):
    merged_list = []
    
    for item in list1:
        merged_list.append(item)
        
    for item in list2:
        merged_list.append(item)
        
    return merged_list
list_a = [1, 2, 3]
list_b = [4, 5, 6]

result = merge_lists(list_a, list_b)
print("Merged list:", result)
```

```python
def convert_to_tuple(input_list):
    def pack_to_tuple(*args):
        return args    
    return pack_to_tuple(*input_list)

my_list = [10, 20, 30, 40, 50]
my_tuple = convert_to_tuple(my_list)

print("List:", my_list)
print("Tuple:", my_tuple)
print("Type:", type(my_tuple))
```

```python
def calculate_sum(numbers):
    total = 0
    for num in numbers:
        total = total + num
    return total

my_list = [10, 20, 30, 40, 50]
result = calculate_sum(my_list)

print("The sum of the list is:", result)
```

```python
def is_sorted(lst):
    for i in range(len(lst) - 1):
        if lst[i] > lst[i + 1]:
            return False
    return True
nums = [1, 2, 3, 4, 5]
print("Is sorted?", is_sorted(nums))
nums2 = [3, 1, 4, 2]
print("Is sorted?", is_sorted(nums2))
```

```python
 def create_dictionary(keys, values):
    result_dict = {}
    for i in range(len(keys)):
        key = keys[i]
        value = values[i]
        result_dict[key] = value       
    return result_dict
list_keys = ["name", "age", "city"]
list_values = ["Alice", 25, "New York"]

my_dict = create_dictionary(list_keys, list_values)
print(my_dict)
```

```python
def count_word_frequency(text):
    words = text.split()
    frequency = {}
    for word in words:
        word = word.lower()        
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
            
    return frequency

sample_text = "Apple Banana apple Cherry Banana apple"
result = count_word_frequency(sample_text)

print(result)
```

```python
def remove_key_with_del(my_dict, key_to_remove):
    if key_to_remove in my_dict:
        del my_dict[key_to_remove]
    return my_dict

person = {"name": "Alice", "age": 25, "city": "New York"}
result = remove_key_with_del(person, "age")

print(result)
```

```python
def sort_dict_manual(d):
    items = list(d.items())
    n = len(items)
    
    for i in range(n):
        for j in range(0, n - i - 1):
            if items[j][1] > items[j + 1][1]:
                items[j], items[j + 1] = items[j + 1], items[j]
                
    return dict(items)

my_dict = {"apple": 10, "orange": 5, "banana": 20, "cherry": 15}
result = sort_dict_manual(my_dict)

print(result)
```

```python
def find_intersection(set1, set2):
    intersection_result = set()
    
    for item in set1:
        if item in set2:
            intersection_result.add(item)
            
    return intersection_result

s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

result = find_intersection(s1, s2)
print(result)
```

```python
def find_union(set1, set2):
    union_result = set()
    
    for item in set1:
        union_result.add(item)
        
    for item in set2:
        union_result.add(item)
        
    return union_result

s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

result = find_union(s1, s2)
print("Union of sets:", result)
```

```python
def is_subset(small_set, big_set):
    for item in small_set:
        if item not in big_set:
            return False
    return True

set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}

print(f"Is set_a a subset of set_b? {is_subset(set_a, set_b)}")
```

```python
def find_common_three(list1, list2, list3):
    common_elements = []
    
    for item in list1:
        if item in list2 and item in list3:
            if item not in common_elements:
                common_elements.append(item)
                
    return common_elements

a = [1, 5, 10, 20, 40, 80]
b = [6, 7, 20, 80, 100]
c = [3, 4, 15, 20, 30, 70, 80, 120]

result = find_common_three(a, b, c)
print(result)
```
