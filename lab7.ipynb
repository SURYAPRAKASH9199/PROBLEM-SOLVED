```python
import pandas as pd

df = pd.read_csv(r"C:\Users\KIIT0001\Downloads\car data.csv")

for col in df.columns:
    if df[col].dtype in ['float64', 'int64']:
        df[col] = df[col].fillna(df[col].mean())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])

print(df)
```

```python
import pandas as pd

df1 = pd.DataFrame({'Car_Name': ['ritz', 'sx4', 'ciaz'], 'Owner': [0, 1, 0]})
df2 = pd.DataFrame({'Car_Name': ['ritz', 'sx4', 'wagon r'], 'Rating': [4, 3, 5]})

inner_join = pd.merge(df1, df2, on='Car_Name', how='inner')
outer_join = pd.merge(df1, df2, on='Car_Name', how='outer')
left_join = pd.merge(df1, df2, on='Car_Name', how='left')
right_join = pd.merge(df1, df2, on='Car_Name', how='right')

print(inner_join)
print(outer_join)
print(left_join)
print(right_join)
```

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\KIIT0001\Downloads\car data.csv")

Q1 = df['Selling_Price'].quantile(0.25)
Q3 = df['Selling_Price'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Selling_Price'] < lower_bound) | (df['Selling_Price'] > upper_bound)]

print(outliers)
```
